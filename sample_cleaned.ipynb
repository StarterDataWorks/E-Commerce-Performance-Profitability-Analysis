```python
df=df.drop_duplicates() # elimino le righe duplicate
```

```python
#Calcola il costo stimato dei prodotti  e il margine %
df['Cost']=(df['Sales']-df['Profit']).round(2)
df['Profit_Margin_%']=(df['Profit']/df['Sales']).round(4)
df['Sales']=df['Sales'].round(2)
df['Profit']=df['Profit'].round(2)
```
